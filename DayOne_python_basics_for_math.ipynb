Prime of N below 50

```python
def is_prime(n):
  if n < 2:
    return False
  for d in range(2, n):
    if n % d == 0:
      return False
  return True
prime_between_50_and_100 = []
for i in range(50,101):
  if is_prime(i):
    prime_between_50_and_100.append(i)
print(prime_between_50_and_100)
```

Finding FIB(N)

```python
def fib(n):
  if n == 0 or n == 1:
    return n
  return fib(n-1) + fib(n-2)

fib_list = []
for i in range(10):
  fib_list.append(fib(i))
print(fib_list)
```

MEAN VALUES

```python
def mean(values: list) -> float:
  total = 0
  for i in values:
    total += i
  return total / len(values)
mean([1,2,3])
```

. How large must n be to get 3 correct decimals?

```python
def approximate_pi(n):
  return 4 * sum(-1**k / (2*k + 1) for k in range(n + 1))
print(f"Approximation at n=3999: {approximate_pi(3999):.4f}")
```

2. Plotting the Bell CurveTo plot $f(x) = e^{-x^2}$ (the bell curve shape) on the interval $[-3, 3]$ using a list-based approach

```python
import matplotlib.pyplot as plt
import math

x_vals = [x / 100 for x in range(-300, 301)]
y_vals = [math.exp(-x**2) for x in x_vals]

plt.plot(x_vals, y_vals, color='#D4AF37')
plt.title("f(x) = e^(-x^2)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, alpha=0.3)
plt.show()
```

3. Estimating the DerivativeThe function deriv(f, x, h=1e-5) estimates the derivative $f'(x)$ numerically using the difference quotient $\frac{f(x+h) - f(x)}{h}$. Testing this on the polynomial $f(x) = x^2$ at $x = 3$ yields a result extremely close to the true analytical derivative of 6.  

```python
def deriv(f, x, h=1e-5):
  return (f(x+h)-f(x))/h
slope = deriv(lambda x: x**2, 3)
print(f"Estimated derivative: {slope}")
```

4. Computing the Harmonic Number
The harmonic number $H_n = \sum_{k=1}^n 1/k$ evaluated at $n = 1, 10, 100, 1000$ demonstrates highly logarithmic, slow growth.

```python
def harmonic_number(n):
  return sum(1/k for k in range(1, n + 1))
n_values = [1, 10, 100, 1000]
for n in n_values:
  print(f"H_{n:<4} = {harmonic_number(n):.5f}")
```
