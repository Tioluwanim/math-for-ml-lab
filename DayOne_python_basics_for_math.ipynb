Prime of N below 50

```python
def is_prime(n):
  if n < 2:
    return False
  for d in range(2, n):
    if n % d == 0:
      return False
  return True
prime_between_50_and_100 = []
for i in range(50,101):
  if is_prime(i):
    prime_between_50_and_100.append(i)
print(prime_between_50_and_100)
```

Finding FIB(N)

```python
def fib(n):
  if n == 0 or n == 1:
    return n
  return fib(n-1) + fib(n-2)

fib_list = []
for i in range(10):
  fib_list.append(fib(i))
print(fib_list)
```

MEAN VALUES

```python
def mean(values: list) -> float:
  total = 0
  for i in values:
    total += i
  return total / len(values)
mean([1,2,3])
```
